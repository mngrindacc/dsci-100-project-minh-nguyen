```r
missing_summary <- women_tidy|>
summarize(across(year:bid,~sum(is.na(.x)))) 
missing_summary
#to calculate the number of missing values for each column, we use the across function to tell R that we want to apply
#a function to calculate the number of missing values throughout all the columns. The argument is.na(.x), where
#.x represents each individual column, will output TRUE or FALSE depending on whether
#a row contains missing values. When is.na() is passed into the sum function, R
#converts TRUE values to 1 and FALSE values to 0 then calculates the sum of all the outputs, giving us the total number 
#of missing values. By using the across function and passing the year:bid argument(all columns in our data)
#into it, we are telling R that we want to apply the sum and is.na() functions to all columns. 
#Finally, by using summarize, R outputs a data frame containing the summary statistics
#for the column range stated in across. 
```

This scatter plot directly relates to our research question by showing how well conference and regional win percentages distinguish between different bid types. Our plot shows that as conference win percentages increase, regional wins also increase, revealing a  positive trend in our data. However, despite this positive correlation the bid types still overlap significantly throughout the rest of our data, highlighting that high win percentages alone do not distinguish at large and automatic qualifications and require classification models to capture complex boundaries rather than simple linear thresholds. 

```r
ggplot(women_tidy, aes(x = conf_wins_pct, fill = bid)) +
  geom_histogram() +
  facet_grid(cols=vars(bid)) + #this creates 2 graphs that are split by the variable "bid". 
  labs(x = "Conference Win Percentage", y = "Count", fill = "Bid Type")+#this adds titles to our axes and legends
ggtitle("Histogram showing Conference win percentage distributions 
        for at large and automatic bids")
```

This faceted histogram relates to our research question by evaluating how conference win percentage helps predict bid type. Looking at the graph for at-large bids, we see that conference win percentages concentrate around the 60-80 range. However, for automatic bids, conference win percentages skew towards the higher end. This highlights that higher conference win percentages(80-100%) typically qualify through automatic bids while teams with a 60-80% win percentage are more likely to qualify for at large bids. This graph therefore reveals the direct relationship between conference win percentages and bid types, highlighting this variable as a strong predictor.
